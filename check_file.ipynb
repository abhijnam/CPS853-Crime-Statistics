```python
import pandas as pd
import os
import math
```

```python
try:
    df = pd.read_csv("Crime Data.csv",encoding="latin-1")
    print(df.dtypes)
    file_size_bytes = os.path.getsize("Crime Data.csv")
    file_size_mb = file_size_bytes / (1024 * 1024)
    df.describe()
    print("Size of CSV file",math.ceil(file_size_mb),"MB")
    df.to_parquet('Crime Data.parquet', compression='snappy')
    file_size_bytes_pq = os.path.getsize("Crime Data.parquet")
    file_size_mb_pq = file_size_bytes / (1024 * 1024)
    print("Size of parquet file",math.ceil(file_size_mb_pq),"MB")
except FileNotFoundError:
    print("File not found in directory")
```
