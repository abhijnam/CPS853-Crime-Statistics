```python
import json
from confluent_kafka import Consumer
```

```python


consumer = Consumer({
    "bootstrap.servers": "kafka:9092",
    "group.id": "crime-demo-group-1",
    "auto.offset.reset": "earliest",   # read from the beginning of the topic
})

consumer.subscribe(["crime-events"])

print("Consumer listening for messages...\n")
empty_polls = 0

try:
    while empty_polls < 5:
        msg = consumer.poll(timeout=1.0)
        if msg is None:
            empty_polls += 1
            continue
        if msg.error():
            print("Consumer error:", msg.error())
            continue

        empty_polls = 0
        key = msg.key().decode("utf-8") if msg.key() else None
        event = json.loads(msg.value().decode("utf-8"))
        print(
            f"[Partition {msg.partition()} | Offset {msg.offset()}] "
            f"Key={key} | DR_NO={event['DR_NO']} | "
            f"DATE OCC={event['DATE OCC']} | TIME OCC={event['TIME OCC']} | "
            f"Crime={event['Crm Cd Desc']}"
        )
finally:
    consumer.close()
    print("\nConsumer finished.")
```
